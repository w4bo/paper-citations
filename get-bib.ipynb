```python
from pybliometrics.scopus import AuthorRetrieval, AbstractRetrieval
import bibtexparser
from bibtexparser.bwriter import BibTexWriter
from bibtexparser.bibdatabase import BibDatabase

with open("/home/config.yml") as stream:
    config = yaml.safe_load(stream)
print(config)

def scopus_to_bibtex_type(doc):
    """
    Determine BibTeX type from Scopus document metadata.
    """

    subtype = getattr(doc, "subtypeDescription", "") or ""
    aggregation = getattr(doc, "aggregationType", "") or ""

    subtype = subtype.lower()
    aggregation = aggregation.lower()

    if "conference" in subtype or aggregation == "conference proceeding":
        return "inproceedings"

    if aggregation == "journal":
        return "article"

    if aggregation == "book":
        return "book"

    if "chapter" in subtype:
        return "incollection"

    return "misc"


def format_authors(authors):

    if not authors:
        return ""

    formatted = []

    for a in authors:
        surname = getattr(a, "surname", "") or ""
        given = getattr(a, "given_name", "") or ""

        formatted.append(f"{surname}, {given}")

    return " and ".join(formatted)


author = AuthorRetrieval(config["SCOPUS-ID"])

db = BibDatabase()
db.entries = []

for doc in author.get_documents():

    try:

        # Full metadata
        ab = AbstractRetrieval(doc.eid)

        entry_type = scopus_to_bibtex_type(doc)

        entry = {
            "ENTRYTYPE": entry_type,
            "ID": doc.eid.replace("2-s2.0-", ""),
            "title": ab.title or "",
            "author": format_authors(ab.authors),
            "year": ab.coverDate[:4] if ab.coverDate else "",
            "doi": ab.doi or "",
        }

        if entry_type == "article":
            entry["journal"] = ab.publicationName or ""
            entry["volume"] = getattr(ab, "volume", "") or ""
            entry["number"] = getattr(ab, "issueIdentifier", "") or ""
            entry["pages"] = getattr(ab, "pageRange", "") or ""

        elif entry_type == "inproceedings":
            entry["booktitle"] = ab.publicationName or ""
            entry["pages"] = getattr(ab, "pageRange", "") or ""

        elif entry_type == "incollection":
            entry["booktitle"] = ab.publicationName or ""
            entry["pages"] = getattr(ab, "pageRange", "") or ""

        elif entry_type == "book":
            entry["publisher"] = getattr(ab, "publisher", "") or ""

        db.entries.append(entry)

        print(f"Added: {ab.title}")

    except Exception as e:
        print(f"Failed for {doc.eid}: {e}")

writer = BibTexWriter()
writer.indent = "    "

with open("refs.bib", "w", encoding="utf-8") as bibfile:
    bibfile.write(writer.write(db))

print("Saved refs.bib")
```
