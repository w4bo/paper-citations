```python
import pandas as pd
import glob
import os
import numpy as np

def map_id(x):
    if str(x).lower() == "nan":
        return np.nan
    return str(x).lower().replace('.', '').replace(' ', '').replace(':', '').replace('\'', '').replace('-', '').replace('?', '').replace('\xa0', '').replace(',', '').replace('+', '').replace('<sup>', '').replace('</sup>', '').replace('(discussionpaper)', '')


def get_df(s, path = '/home/data/', onlylast=False):
    i = 0
    li = []
    files = glob.glob(os.path.join(path , s + "*.csv"))
    for filename in files if not onlylast else [files[-1]]:
        # print(f"{i} {filename}")
        # i += 1
        df = pd.read_csv(filename, index_col=None, header=0, encoding='utf-8')
        li.append(df)
    all_df = pd.concat(li, axis=0, ignore_index=True)
    all_df = all_df.drop_duplicates().replace("", np.nan)
    all_df["citeid"] = all_df["Paper title"].apply(lambda x: map_id(x))
    all_df["citingid"] = all_df["Citing paper title"].apply(lambda x: map_id(x))
    return all_df
```

```python
scholar_unique = scholar["citeid"].sort_values().unique()
len(scholar_unique)
```

```python
not_cited_ids = []
for x in scholar_unique:  # for each paper in scholar
    citations_scholar = scholar[scholar["citeid"] == x]  # get its citing papers from scholar
    paper_scopus = scopus[(scopus["citeid"] == x)]  # get the paper from scopus...
    if len(paper_scopus) == 0:  # if the paper is not on scopus...
        print(x + " is not in scopus")
    else:
        for y in citations_scholar["citingid"].unique():  # for each of the citing papers from scholar
            # if (citations_scholar["Citing paper title"].str.contains('Compressione').any()): print("Here")
            if not paper_scopus["citingid"].str.contains(y).any():  # if this citation is not in scopus...
                not_cited_ids.append([x, y])  # add it
not_cited = pd.DataFrame(not_cited_ids, columns=["citeid", "citingid"])
not_cited
```

```python
missing_papers = scholar[["Citing paper title", "citingid"]].drop_duplicates().merge(not_cited, on=["citingid"])
missing_papers
```

```python
scopus_acc = []
not_in_scopus = []
missin_papers_scopus = pd.DataFrame(scopus_acc, columns=["Citing paper title", "Citing paper eid", "Citing paper doi", "citingid"])
missin_papers_scopus
```

```python
import pybliometrics
import yaml
with open("/home/config.yml") as stream:
    config = yaml.safe_load(stream)
pybliometrics.scopus.init(config_dir=config["config_dir"])
from pybliometrics.scopus import ScopusSearch
from pybliometrics.scopus import AuthorRetrieval
import sys
import pandas as pd
import time
from random import shuffle
import re

def map_string(x):
    skip_words = [":", "'", " and ", " or ", " of ", " the ", "-"]
    x = x.lower()
    for s in skip_words:
        x = x.replace(s, " ")
    _RE_COMBINE_WHITESPACE = re.compile(r"\s+")
    x = _RE_COMBINE_WHITESPACE.sub(" ", x).strip()
    return x

xx = list(missing_papers[["Citing paper title", "citingid"]].drop_duplicates().to_numpy())
shuffle(xx)
for x in xx: #[["PUTS: Privacy-Preserving and Utility-Enhancing Framework for Trajectory Synthesization", "foo"]]:
    string = "TITLE-ABS-KEY({})".format(' AND '.join(map_string(x[0]).split(' ')))
    try:
        s = ScopusSearch(string)
        # print(string)
        # print(s.results)
        if s.results is not None and len(s.results) > 0:
            papers = [[y.title, y.eid, y.doi, x[1]] for y in s.results]
            for paper in papers:
                scopus_acc = scopus_acc + [paper]
                missin_papers_scopus = pd.DataFrame(scopus_acc, columns=["Citing paper title", "Citing paper eid", "Citing paper doi", "citingid"]).drop_duplicates()
        else:
            not_in_scopus = not_in_scopus + [x[0]]
        time.sleep(1)
    except Exception as e:
        print(x[0] + " " + string + ": " + str(e) + "\n")
```

```python
def filter_col(df):
    df = df[[x for x in df.columns if x != "citingid" and x != "citeid" and "doi" not in x]]
    df = df.rename({"Paper title": "Cited paper title", "Paper eid": "Cited paper eid"}, axis=1)
    df["Cited paper link"] = df["Cited paper eid"].apply(lambda x: "https://www.scopus.com/record/display.uri?eid={}&origin=resultslist".format(x))
    df["Citing paper link"] = df["Citing paper eid"].apply(lambda x: "https://www.scopus.com/record/display.uri?eid={}&origin=resultslist".format(x))
    return df
```

```python
filter_col(scopus_papers.merge(not_cited, on=["citeid"]).merge(missin_papers_scopus, on=["citingid"])).sort_values(by=['Cited paper title', 'Citing paper title']).to_csv('/home/data/missing-{}-all.csv'.format(time.time()), index=False)
filter_col(scopus_papers.merge(not_cited, on=["citeid"]).merge(missin_papers_scopus[missin_papers_scopus["citingid"] == missin_papers_scopus["Citing paper title"].apply(lambda x: map_id(x))], on=["citingid"])).sort_values(by=['Cited paper title', 'Citing paper title']).to_csv('/home/data/missing-{}-red.csv'.format(time.time()), index=False)
```
