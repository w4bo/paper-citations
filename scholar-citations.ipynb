```python
from scholarly import scholarly
import yaml
import time
import pandas as pd

with open("config.yml") as stream:
    config = yaml.safe_load(stream)
```

```python
# Retrieve the author's data, fill-in, and print
# Get an iterator for the author results
search_query = scholarly.search_author('Matteo Francia, University of Bologna')
# Retrieve the first result from the iterator
first_author_result = next(search_query)
scholarly.pprint(first_author_result)
```

```python
# Retrieve all the details for the author
author = scholarly.fill(first_author_result)
scholarly.pprint(author)
```

```python
publications = author['publications']
publications
```

```python
from random import shuffle
shuffle(publications)
publications
```

```python
acc = []
```

```python
try:
    for first_publication in publications:
        if first_publication["num_citations"] > 0:
            time.sleep(3)
            first_publication_filled = scholarly.fill(first_publication)
            time.sleep(2)
            citations = [citation['bib']['title'] for citation in scholarly.citedby(first_publication_filled)]
            for c in citations:
                acc = acc + [[first_publication_filled['bib']['title'], c]]
except Exception as e:
    print(e)
```

```python
if len(acc) > 0:
    pd.DataFrame(acc, columns=["Paper title", "Citing paper title"]).to_csv("data/scholar-citations-{}.csv".format(time.time()), index=False, encoding='utf-8')
```
